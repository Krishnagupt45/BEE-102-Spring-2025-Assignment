```python
def convert_fasta(input_file, output_file):
    with open(input_file, 'r') as fin, open(output_file, 'w') as fout:
        seq = ''
        for line in fin:
            line = line.strip()
            if line.startswith('>'):
                if seq:
                    fout.write(seq + '\n')
                fout.write(line + '\n')
                seq = ''
            else:
                seq += line
        if seq:
            fout.write(seq + '\n')
convert_fasta('input.fasta', 'output.fasta')
```
