```python
import numpy as np
import pandas as pd
def build_markov_matrix(sequence):
    nucleotides = ['A', 'C', 'G', 'T']
    transition_counts = {n1: {n2: 0 for n2 in nucleotides} for n1 in nucleotides}
    for i in range(len(sequence) - 1):
        current = sequence[i]
        next_ = sequence[i + 1]
        if current in nucleotides and next_ in nucleotides:
            transition_counts[current][next_] += 1
    df = pd.DataFrame(transition_counts).T
    transition_matrix = df.div(df.sum(axis=1), axis=0).fillna(0)
    print("Markov Transition Matrix:")
    print(transition_matrix.round(3))

sequence = "ACGTAGCTAGCTAGGGCTA"
build_markov_matrix(sequence)
```
