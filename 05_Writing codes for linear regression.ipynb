```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression

area_sqft = np.array([500, 750, 1000, 1250, 1500, 1750, 2000, 2250, 2500, 2750]).reshape(-1, 1)
price_lakhs = np.array([30, 42, 55, 65, 78, 82, 98, 108, 115, 125])
model = LinearRegression()
model.fit(area_sqft, price_lakhs)
predicted_prices = model.predict(area_sqft)

plt.scatter(area_sqft, price_lakhs, color='blue', label='Actual')
plt.plot(area_sqft, predicted_prices, color='red', label='Predicted')
plt.xlabel('Plot Area (sq ft)')
plt.ylabel('Price (₹ lakhs)')
plt.title('Linear Regression: Plot Area vs Price')
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()

mean_price = np.mean(price_lakhs)
TSS = np.sum((price_lakhs - mean_price)**2)
ESS = np.sum((predicted_prices - mean_price)**2)
RSS = np.sum((price_lakhs - predicted_prices)**2)

print(f"Total Sum of Squares (TSS): {TSS:.2f}")
print(f"Explained Sum of Squares (ESS): {ESS:.2f}")
print(f"Residual Sum of Squares (RSS): {RSS:.2f}")
```
