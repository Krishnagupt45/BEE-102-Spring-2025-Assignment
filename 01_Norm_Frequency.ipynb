```python
import pandas as pd
import matplotlib.pyplot as plt
import gzip
data = pd.read_csv('query.bed.gz', sep='\t', header=None, compression='gzip')
data['fragment_length'] = data[2] - data[1]
counts = data['fragment_length'].value_counts().sort_index()
normalized_counts = counts / counts.sum()
plt.figure(figsize=(10,6))
plt.plot(normalized_counts.index, normalized_counts.values)
plt.xlabel('Fragment Length (bp)')
plt.ylabel('Normalized Frequency [A.U.]')
plt.title('Normalized Fragment Length Frequency')
plt.grid(True)
plt.show()
```
